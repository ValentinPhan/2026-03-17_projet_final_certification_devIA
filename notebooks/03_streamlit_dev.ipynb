```python
import joblib
import pandas as pd

model = joblib.load("../models/rf_gravite_pipeline.joblib")

# Exemple de saisie utilisateur simulée
test_input = pd.DataFrame([{
    "annee": 2022,
    "type_evenement": "collision",
    "exploitant": "SNCF Réseau",
    "departement": 69,
    "cause_presumee": "erreur_humaine",
    "contexte": "circulation_commerciale",
    "source": "BEATT_rapport"
}])

pred = model.predict(test_input)
proba = model.predict_proba(test_input)

label_map = {0: "Mineur", 1: "Significatif", 2: "Grave"}
# print("Prédiction :", label_map[pred[0]])
print("Prédiction :", pred[0].capitalize())
print("Probabilités :", dict(zip(["Mineur", "Significatif", "Grave"], proba[0].round(2))))
```
